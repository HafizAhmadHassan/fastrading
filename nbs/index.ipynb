````python
import io
from gradio_client import Client, handle_file
import ipywidgets as widgets
from IPython.display import display, clear_output

# Initialize the Gradio client
client = Client("HafizAhmadHassan/dog-with-cat")

# Create widgets
file_upload = widgets.FileUpload(accept='image/*', multiple=False)
predict_button = widgets.Button(description="Predict", button_style='success')
output = widgets.Output()

# Define the prediction function
def predict_image(change):
    with output:
        clear_output()
        if file_upload.value:
            # Extract the uploaded file
            uploaded_file = file_upload.value[0]
            file_name = uploaded_file['name']
            file_bytes = uploaded_file['content']
            
            # Save the uploaded file to a temporary location
            with open(file_name, 'wb') as f:
                f.write(file_bytes)
            
            # Use handle_file to prepare the file for prediction
            result = client.predict(
                img=handle_file(file_name),
                api_name="/predict"
            )
            
            # Display the prediction result
            print("Prediction Result:")
            print(result)
        else:
            print("Please upload an image file.")

# Attach the prediction function to the button click event
predict_button.on_click(predict_image)

# Display the widgets
display(widgets.VBox([file_upload, predict_button, output]))
````

````python
import gradio as gr
from gradio_client import Client

# Initialize your client (replace with your actual model link if needed)
client = Client("HafizAhmadHassan/dog-with-cat")

def predict_image(image):
    result = client.predict(
        img=image,
        api_name="/predict"
    )
    return result

# Create the interface
interface = gr.Interface(
    fn=predict_image,
    inputs=gr.Image(type="filepath", label="Upload an image"),
    outputs=gr.Label(label="Prediction"),
    title="Dog-with-Cat Detector",
    description="Upload an image to see prediction using FastHTML API."
)

# Launch the interface
interface.launch()
````
