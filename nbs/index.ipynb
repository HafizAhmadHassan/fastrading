````python
import gradio as gr
from gradio_client import Client

# Initialize your client (replace with your actual model link if needed)
client = Client("HafizAhmadHassan/dog-with-cat")

def predict_image(image):
    result = client.predict(
        img=image,
        api_name="/predict"
    )
    return result

# Create the interface
interface = gr.Interface(
    fn=predict_image,
    inputs=gr.Image(type="filepath", label="Upload an image"),
    outputs=gr.Label(label="Prediction"),
    title="Dog-with-Cat Detector",
    description="Upload an image to see prediction using FastHTML API."
)

# Launch the interface
interface.launch()
````
